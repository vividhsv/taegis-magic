```python
import re

pattern = "|".join(rf"(^|\.){re.escape(d)}$" for d in domains)

query = f"""
WITH UniqueHosts AS (
    SELECT DISTINCT uri_host FROM http
),
MatchedHosts AS (
    SELECT uri_host FROM UniqueHosts
    WHERE regexp_like(uri_host, '{pattern}')
)
SELECT h.*
FROM http h
JOIN MatchedHosts m ON h.uri_host = m.uri_host
"""
```
